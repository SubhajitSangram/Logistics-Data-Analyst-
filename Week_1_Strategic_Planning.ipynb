## Objective
Define the logistics scenario, KPIs and strategic analytical roadmap.

```python
import pandas as pd
from pathlib import Path

DATA = Path("data/logistics_master_raw.csv")
df = pd.read_csv(DATA, parse_dates=["order_date"])

print(df.shape)
display(df.head())
```

```python
df["late_flag"] = (df["delivery_days"] > df["promised_days"]).astype(int)

kpis = {
    "On-time delivery rate (%)": (1 - df["late_flag"].mean()) * 100,
    "Late delivery rate (%)": df["late_flag"].mean() * 100,
    "Average delivery days": df["delivery_days"].mean(),
    "Average delay days": df["delay_days"].mean(),
    "Freight cost/order": df["freight_cost"].mean(),
    "Stockout rate (%)": df["stockout_flag"].mean() * 100,
}
pd.Series(kpis).round(2)
```

## Strategic roadmap
Business understanding → data validation → cleaning → EDA → feature engineering → prediction → segmentation → optimization → monitoring.
