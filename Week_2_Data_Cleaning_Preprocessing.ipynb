## Load intentionally imperfect data

```python
import pandas as pd
import numpy as np
from pathlib import Path

df = pd.read_csv("data/week2_dirty_logistics.csv")
print(df.shape)
print(df.isna().sum())
```

## Date conversion and missing-value handling

```python
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")

df["weight_kg"] = df["weight_kg"].fillna(df["weight_kg"].median())
df["freight_cost"] = df["freight_cost"].fillna(df["freight_cost"].median())
df["transport_mode"] = df["transport_mode"].fillna(df["transport_mode"].mode()[0])

print("Missing values after basic imputation:")
print(df.isna().sum())
```

## Outlier detection using IQR

```python
q1 = df["distance_km"].quantile(0.25)
q3 = df["distance_km"].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

df["distance_outlier"] = (df["distance_km"] < lower) | (df["distance_km"] > upper)

print("IQR bounds:", lower, upper)
print(df["distance_outlier"].value_counts())
```

## Save cleaned data

```python
clean = df.drop(columns=["distance_outlier"]).copy()
clean.to_csv("data/week2_cleaned_from_notebook.csv", index=False)
print("Saved cleaned dataset.")
```
